Access data lake using access keys
1. Register azure AD application/service principal 
2. Generate a secret/password for the application
3. Set spark config with App/client id, directory/Tenant id and secrets
4. Assign role "Storage blob data contributor" to data lake

```python
dbutils.secrets.list(scope='formula1-scope')
```

```python
client_secret=dbutils.secrets.get(scope='formula1-scope', key='service-account-kv')
```

```python
client_id="aeddf8bb-d4c2-4c2f-8cff-6f2ff7542040"
teanant_id="c172ff36-4771-411e-80fe-e795b9c3e0f0"
client_secret=client_secret
```

```python

spark.conf.set(
    "fs.azure.account.auth.type.formula78dll.dfs.core.windows.net",
    "OAuth"
)

spark.conf.set(
    "fs.azure.account.oauth.provider.type.formula78dll.dfs.core.windows.net",
    "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider"
)

spark.conf.set(
    "fs.azure.account.oauth2.client.id.formula78dll.dfs.core.windows.net",
    client_id  
)

spark.conf.set(
    "fs.azure.account.oauth2.client.secret.formula78dll.dfs.core.windows.net",
    client_secret 
)

spark.conf.set(
    "fs.azure.account.oauth2.client.endpoint.formula78dll.dfs.core.windows.net",
    f"https://login.microsoftonline.com/{teanant_id}/oauth2/token"
)
```
