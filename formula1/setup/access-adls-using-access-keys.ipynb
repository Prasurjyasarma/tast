Access data lake using access keys
1. Set the spark config account key 
2. List files from demo container
3. Read data from circuits.csv file

```python
formula1_dl_accountkey=dbutils.secrets.get(scope='formula1-scope', key='formula1-dl-account-key')
```

```python
spark.conf.set( "fs.azure.account.key.formula78dll.dfs.core.windows.net",
               formula1_dl_accountkey)
```

```python
dbutils.fs.ls("abfss://demo@formula78dll.dfs.core.windows.net/")
```

```python
spark.read.csv("abfss://demo@formula78dll.dfs.core.windows.net/circuits.csv")
```
