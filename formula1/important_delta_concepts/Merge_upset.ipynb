```python
from pyspark.sql.functions import expr
from delta.tables import DeltaTable
```

```python
deltaTable = DeltaTable.forName(spark, "api_formula1.default.circuits_table") # <- t
```

```python
updatesDF = spark.createDataFrame(
    [
        (3, "bahrain", "Bahrain International Circuit", "Sakhir", "Kingdom of Bahrain", 26.0325, 50.5106, 7, "2025-08-27T15:47:35.384+00:00", "2025-08-27"),
        (10, "vegas", "Las Vegas Grand Prix Circuit", "Las Vegas", "USA", 36.1699, -115.1398, 620, "2025-08-27T15:47:35.384+00:00", "2025-08-27")
    ],
    ["circuit_id","circuit_ref","name","location","country","latitude","longitude","altitude","ingestion_date","file_date"]
)        # <- s
```

```python
deltaTable.alias("t").merge(
    updatesDF.alias("s"),
    "t.circuit_id = s.circuit_id"
).whenMatchedUpdate(set = {      
    "circuit_ref": "s.circuit_ref",
    "name": "s.name",
    "location": "s.location",
    "country": "s.country",
    "latitude": "s.latitude",
    "longitude": "s.longitude",
    "altitude": "s.altitude",
    "ingestion_date": "s.ingestion_date",
    "file_date": "s.file_date"
}).whenNotMatchedInsert(values = {
    "circuit_id": "s.circuit_id",
    "circuit_ref": "s.circuit_ref",
    "name": "s.name",
    "location": "s.location",
    "country": "s.country",
    "latitude": "s.latitude",
    "longitude": "s.longitude",
    "altitude": "s.altitude",
    "ingestion_date": "s.ingestion_date",
    "file_date": "s.file_date"
}).execute()
```

```python
df=deltaTable.toDF()
```
